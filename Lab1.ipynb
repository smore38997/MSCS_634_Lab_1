```python
import pandas as pd 


df = pd.read_csv("mobs.csv")

df.head()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["type"], df["height"])

plt.xlabel("type")
plt.ylabel("height")
plt.title("Mob Type Vs Height ")

plt.show()
```

```python
plt.hist(df["height"], bins=10)


plt.title("Distribution of Mob Height")

plt.show()
```
